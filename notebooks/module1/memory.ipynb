<h2>Without Memory</h2>

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.messages import HumanMessage

model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")

agent=create_agent(
    model=model
)

response=agent.invoke(
    {'messages':[HumanMessage(content="My name is Romana and my favourite breakfast is Dosa")]}
)

print(response)
```

```python
Question=HumanMessage(content="What is my favourite breakfast?")
response=agent.invoke(
    {'messages':[Question]}
)

print(response['messages'][-1].content)
```

<h2>With Memory</h2>

```python
from langgraph.checkpoint.memory import InMemorySaver

agent=create_agent(
    model=model,
    checkpointer=InMemorySaver(),
)
```

```python
from langchain.messages import HumanMessage

Question=HumanMessage(content="My name is Romana and my favourite breakfast is Dosa")
config ={"configurable":{"thread_id":"1"}}

response=agent.invoke(
    {'messages':[Question]},
    config,
)

print(response['messages'][-1].content)
```

```python
question = HumanMessage(content="What's my favourite breakfast?")

resp = agent.invoke(
    {"messages": [question]},
    config,  
)


print(resp['messages'][-1].content[0]['text'])
```
