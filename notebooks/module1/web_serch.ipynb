<h2>Without Web Search Tool</h2>

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.messages import HumanMessage
from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-20b",
    
)
question=HumanMessage(content="How up to date your training knowledge? And when was TVK party formed in Tamilnadu and who formed it")

agent=create_agent(
    model=model
)

response =agent.invoke(
    {'messages':[question]}
)
print(response['messages'][-1].content)
```

<h2>Adding Web search Tool</h2>

```python
from langchain.tools import tool
from typing import Dict,Any
from tavily import TavilyClient

tavily_Search = TavilyClient()

@tool
def web_search(query:str) -> Dict[str,Any]:
    """Searching in web for information"""
    return tavily_Search.search(query)

web_search.invoke("when was TVK party formed in Tamilnadu and who formed it?")
```

```python
agent=create_agent(
    model=model,
    tools=[web_search]
)

question = HumanMessage(content="Which Team won T20 mens World Cup this year?")

response=agent.invoke(
    {"messages":[question]}
)


print(response['messages'][-1].content)
```
