<h2>Basic Promting</h2>

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage 
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
agent=create_agent(model=model)
question= HumanMessage(content="what is the nick name for mango?")
response=agent.invoke(
    {"messages":[question]}
)

print(response['messages'][1].content)
```

```python
system_promt="you are fiction writer, create new shot nick name for fruits based on user request."

nick_name_agent=create_agent(
    model=model,
    system_prompt=system_promt
)
response=nick_name_agent.invoke(
    {"messages":[question]}
)

print(response['messages'][1].content)
```

<h2>Few shot promting </h2>

```python
system_promt="""
you are a nick name fiction writer, create a shot nickname for fruits based on user request

user: what is nickname for mango?
nick_name_agent: nick name for mango is mazza

user: what is nickname for watermelon?
nick_name_agent : nick name for watermelon is melonee

"""
nick_name_agent=create_agent(
    model=model,
    system_prompt=system_promt
)
question1= HumanMessage(content="what is nick name for muskmealon")
response=nick_name_agent.invoke(
    {"messages":[question1]}
)

print(response['messages'][1].content[0]["text"])

```

<h2>structured promts </h2>

```python
system_prompt1 = """

you are a nick name fiction writer, create a shot nickname for fruits based on user request

Please keep to the below structure.

Actual Name: The name of the fruit

nick name : the nick name of the fruit

Location: Where it is originated


"""
nick_name_agent=create_agent(
    model=model,
    system_prompt=system_prompt1
)
question1= HumanMessage(content="what is nick name for muskmealon")
response=nick_name_agent.invoke(
    {"messages":[question1]}
)

print(response['messages'][1].content[0]["text"])
```
