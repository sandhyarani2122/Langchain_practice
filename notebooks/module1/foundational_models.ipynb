```python
from dotenv import load_dotenv
load_dotenv()
```

<h1>Initialising and invoking a model<h1>

```python
from langchain.chat_models import init_chat_model
model=init_chat_model(model='gemini-3.6-flash', model_provider="google_genai")
response=model.invoke("what are the colours in rainbow?")
print(response.content)
```

```python
from pprint import pprint
pprint(response.response_metadata)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.6-flash", temperature=1.0)

response = model.invoke("What's the capital of the Moon?")
print(response.content)
```

```python
from langchain_core.messages import HumanMessage

response = model.invoke(
    [HumanMessage(content="What's the capital of the Moon?")]
)

print(response.content)
```

<h1>intialising and invoking an agent</h1>

```python
from langchain.agents import create_agent
# it will take previously intialized model
agent=create_agent(model=model) 

response=agent.invoke(
    {'messages':[HumanMessage(content="what is capital of india")]}
)
pprint(response)
pprint(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="what is capital of india"),
                 AIMessage(content="The capital of India is **New Delhi**."),
                 HumanMessage(content="Tell more about why its made as capital of india")]}
)
pprint(response)
```

<h1>Streaming output </h1>

```python
for token, metadata in agent.stream(
    {'messages':[HumanMessage(content="Tell more about why its made as capital of india")]},
    stream_mode="messages"
    ):
   if token.content:
        for block in token.content:
            if block["type"] == "text":
                print(block["text"], end="", flush=True)

```
