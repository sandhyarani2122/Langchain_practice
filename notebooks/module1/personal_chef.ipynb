<h2>Personal Chef</h2>
<p>An Agent which will take ingrediants and search the web and give you the receipe and answer follow up questions</p>

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.checkpoint.memory import InMemorySaver
from langchain.tools import tool
from tavily import TavilyClient

model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
tavily_search= TavilyClient()
Question= HumanMessage(content="can you give a breakfast recipe with bread, butter, peanutbutter and banana,milk")
config ={"configurable":{"thread_id":"1"}}
@tool
def search_receipe(ingredients:list[str]):
    """Based on list of ingredients it will serach for a recipe"""
    query_s='recipe using'+",".join(ingredients)
    return tavily_search.search(query=query_s)


agent=create_agent(
    model=model,
    tools=[search_receipe],
    checkpointer=InMemorySaver()
)

response=agent.invoke(
    {'messages':[Question]},
    config

)

print(response['messages'][-1].content[0]['text'])
```
